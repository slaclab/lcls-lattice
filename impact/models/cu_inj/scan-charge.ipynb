# Scan `change_timestep_1:dt`

```python
%%time
def set_dt(dt):
    I = Impact.from_yaml('ImpactT.yaml')
    I['change_timestep_1:dt'] = dt
  #  I.stop = 1
    I.run()
    return I
    
set_dt(1e-12)
```

# Parallel

```python
from concurrent.futures import ProcessPoolExecutor
```

```python
# Scan in powers of 2
dtlist = [100e-12 / 2**n for n in range(0,5) ] 
dtlist 
```

```python
%%time
with ProcessPoolExecutor() as executor:
    results = list(executor.map(set_dt, dtlist))
```

```python
fig, ax = plt.subplots()

k1 = 'mean_z'
k2 = 'norm_emit_x'

f1 = 1
f2 = 1e6

for dt, I in zip(dtlist, results):
    
    x = I.stat(k1)*f1
    y = I.stat(k2)*f2
    print(dt, len(y))
    ax.plot(x, y, label=dt)
    
ax.set_ylim(0, 2)    
ax.legend()    
    
```
