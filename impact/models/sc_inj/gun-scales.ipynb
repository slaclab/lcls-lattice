# Gun voltage scaling for `sc_inj` Impact-T

This produces a table for scaling the Impact-T fieldmap to produce a particular kinetic energy.

```python
import numpy as np
from scipy.optimize import brentq
from impact import Impact


IMPACT_IN = '$LCLS_LATTICE/impact/models/sc_inj/v1/ImpactT.in'
def set1(x):
    I = Impact(IMPACT_IN)
    I['RFGUNB']['rf_field_scale'] = x
    I.stop = 0.1
    I.track1()
    return I.stat('mean_kinetic_energy')[-1] 

xlist = np.arange(600e3, 760e3, 5e3)
ylist = []
for target in xlist:
    def f(x):
        return set1(x) - target
    scale = brentq(f, 15e6, 20e6)
    ylist.append(scale)
ylist = np.array(ylist)
```

```python
for x, y in zip(xlist, ylist):
    print(f'{x} {y}')
```
